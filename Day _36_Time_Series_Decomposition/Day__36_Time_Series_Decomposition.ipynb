## 4. Conceptual Overview and Mathematical Intuition

### 4.1 What is Time Series Decomposition?

Time series decomposition is the process of separating a time series into several components:
- **Trend:** The long-term progression of the series (upward, downward, or stable).
- **Seasonality:** Regular, periodic fluctuations (e.g., monthly, quarterly patterns).
- **Residual (Irregular/Noise):** Random fluctuations that cannot be explained by trend or seasonality.

### 4.2 Why Decompose a Time Series?

- **Forecasting:** Understanding the components can improve forecasting accuracy.
- **Anomaly Detection:** Isolating the residuals helps detect outliers or irregular events.
- **Interpretability:** Visualizing the components makes it easier to interpret the data.

### 4.3 Mathematical Intuition

One common model is the **additive model**:
\[
y(t) = T(t) + S(t) + R(t)
\]
where:
- \( y(t) \) is the observed value at time \( t \).
- \( T(t) \) is the trend component.
- \( S(t) \) is the seasonal component.
- \( R(t) \) is the residual component.

For multiplicative models, the components multiply:
\[
y(t) = T(t) \times S(t) \times R(t)
\]
The additive model works well when the seasonal fluctuations are roughly constant through time, while the multiplicative model is suitable when seasonal variations change proportionally with the level of the time series.

## 5. Implementation

Below, we perform time series decomposition on the AirPassengers dataset using Statsmodels.



```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# Load the AirPassengers Dataset

# The AirPassengers dataset is available from a URL
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url, parse_dates=["Month"], index_col="Month")
df.head()
```

**Explanation**  
- We load the AirPassengers dataset from a public URL.
- The "Month" column is parsed as dates and set as the index.
- We display the first few rows to verify the data.


```python
# Exploratory Data Analysis (EDA)

# Plot the original time series
plt.figure(figsize=(12, 6))
plt.plot(df.index, df["Passengers"], label="Passengers")
plt.title("Monthly Number of Airline Passengers")
plt.xlabel("Date")
plt.ylabel("Number of Passengers")
plt.legend()
plt.show()
```

**Explanation**  
- We plot the original time series data to visualize the overall pattern.
- The plot shows the monthly number of passengers, revealing an upward trend and clear seasonality.


```python
# Time Series Decomposition

# Decompose the time series using an additive model

result = seasonal_decompose(df["Passengers"], model="additive", period=12)
result.plot()
plt.tight_layout()
plt.show()
```

**Explanation**  
- We decompose the time series using **seasonal_decompose** with an additive model.
- The `period=12` parameter indicates that the seasonal cycle is 12 months.
- The decomposition result is plotted, showing the trend, seasonal, and residual components.


```python
# Interpretation of Decomposed Components

# Extract decomposed components
trend = result.trend
seasonal = result.seasonal
residual = result.resid

# Plot each component separately for detailed analysis
plt.figure(figsize=(14, 8))

plt.subplot(311)
plt.plot(trend, label="Trend", color="blue")
plt.title("Trend Component")
plt.legend()

plt.subplot(312)
plt.plot(seasonal, label="Seasonality", color="green")
plt.title("Seasonal Component")
plt.legend()

plt.subplot(313)
plt.plot(residual, label="Residuals", color="red")
plt.title("Residual Component")
plt.legend()

plt.tight_layout()
plt.show()
```

**Explanation**  
- We extract the trend, seasonal, and residual components from the decomposition result.
- Separate plots for each component provide a clearer understanding:
  - **Trend:** Shows the overall increase in passenger numbers over time.
  - **Seasonality:** Reveals recurring patterns throughout each year.
  - **Residual:** Captures random fluctuations not explained by the trend or seasonality.
